```python
import time
import gymnasium as gym
from gymnasium.wrappers import RecordVideo

# ---- Build & Wrap Environment ----
base_env = MapMiniGrid(
    json_path=map_path,
    agent_view_size=7,
    render_mode="rgb_array",
)

env = RecordVideo(
    base_env,
    video_folder="videos",
    episode_trigger=lambda ep: True
)

print("[Env] Resetting...")
obs, info = env.reset(seed=0)

done = False
steps = 0

while not done and steps < 100:
    get_next_plan_text = "go to work"
    current_location_text = "home_A"

    # IMPORTANT: use unwrapped env for state access
    path = get_next_plan_and_waypoint(
        current_location_text,
        get_next_plan_text,
        env.unwrapped
    )

    for action in path:
        print(action)
        obs, reward, term, trunc, info = env.step(action)

        # Render forces frame capture for RecordVideo
        frame = env.render()

        vis = get_obs(env.unwrapped, include_world_coords=True, include_street=False)

        time.sleep(0.2)

    done = term or trunc
    steps += 1

env.close()
```
