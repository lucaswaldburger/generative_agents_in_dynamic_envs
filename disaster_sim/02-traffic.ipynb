```python
from environments import *
from globals import *

# register(
#     id="MiniGrid-TrafficSim-v0",
#     entry_point=__name__ + ":TrafficSimEnv",
#     max_episode_steps=100,
#     kwargs={
#         "size": 13,
#         "static_disappearing_mode": False
#     },
# )

register(
    id="MiniGrid-StaticDisappear-v0",
    entry_point=__name__ + ":TrafficSimEnv",
    max_episode_steps=100,
    kwargs={
        "size": 13,
        "static_disappearing_mode": True,
        "disappear_prob": 0.15,
        # "traffic_concentration_pct": 1.0,
        # "traffic_concentration_align": 0.8,
        "target_concentration_x": 10,
        "concentration_span": 3
    },
)

env = gym.make("MiniGrid-StaticDisappear-v0", render_mode="human")
obs, info = env.reset()
print("Mission:", env.unwrapped.mission) 
terminated = False
truncated = False

total_reward = 0

MAX_STEPS = 50_000
SPEED = 0.25

for i in range(MAX_STEPS):
    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(action)

    print(f"Step {i}: Action={action}, Reward={reward:.2f}, Terminated={terminated}, Truncated={truncated}")

    total_reward += reward
    time.sleep(SPEED) 

    if terminated or truncated:
        break

if env.render_mode == "human" and not (terminated or truncated):
    print("Successfully reached the goal")
    time.sleep(SPEED)
    
env.close()
```
