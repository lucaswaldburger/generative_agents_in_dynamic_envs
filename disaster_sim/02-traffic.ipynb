```python
register(
    id="MiniGrid-StaticDisappear-v0",
    entry_point=__name__ + ":TrafficSimEnv",
    max_episode_steps=100,
    kwargs={
        "size": 10,
        "static_disappearing_mode": True,
        "disappear_prob": 0.1,
        # "traffic_concentration_pct": 1.0,
        "traffic_concentration_align": 0.8,
        # "target_concentration_x": 0.0,
        "concentration_span": 10
    },
)

base_env = gym.make("MiniGrid-StaticDisappear-v0", 
                    render_mode="rgb_array",
                    size=10) 

env = RecordVideo(
    base_env, 
    video_folder="traffic_avoidance_videos", 
    episode_trigger=lambda x: True,
    name_prefix="traffic_avoidance_smart_agent",
    fps=5
)

obs, info = env.reset()

NUM_EPISODES = 5
for episode in range(NUM_EPISODES):
    print(f"\n--- Starting Episode {episode + 1} ---")
    for i in range(MAX_STEPS):
        action = env.unwrapped._smart_action_choice()
        obs, reward, terminated, truncated, info = env.step(action)
        
        current_view_size = info.get('view_size', env.unwrapped.agent_view_size)
        print(f"E{episode+1}, Step: {i+1:02d}, Action: {action}, Reward: {reward:.4f}, View Size: {current_view_size}x{current_view_size}, Terminated: {terminated}")
        
        if terminated or truncated:
            print(f"Episode Ended at step {i+1}. Final Reward: {reward:.4f}")
            obs, info = env.reset() 
            break

env.close()
```
