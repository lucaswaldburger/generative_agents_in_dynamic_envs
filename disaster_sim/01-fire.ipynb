```python
register(
    id="MiniGrid-FireSmokeEnv-v0",
    entry_point=__name__ + ":FireSmokeEnv",
)

base_env = gym.make("MiniGrid-FireSmokeEnv-v0", 
                    render_mode="rgb_array",
                    size=16) 

env = RecordVideo(
    base_env, 
    video_folder="fire_avoidance_videos", 
    episode_trigger=lambda x: True,
    name_prefix="fire_avoidance_smart_agent",
    fps=5
)

obs, info = env.reset()
print("\n--- Smart Agent Simulation Started (Fire and Smoke) ---")
print("Recording video to 'fire_avoidance_videos/'...")

NUM_EPISODES = 2
for episode in range(NUM_EPISODES):
    print(f"\n--- Starting Episode {episode + 1} ---")
    for i in range(MAX_STEPS):
        action = env.unwrapped._smart_action_choice()
        obs, reward, terminated, truncated, info = env.step(action)
        
        current_view_size = info.get('view_size', env.unwrapped.agent_view_size)
        print(f"E{episode+1}, Step: {i+1:02d}, Action: {action}, Reward: {reward:.4f}, View Size: {current_view_size}x{current_view_size}, Terminated: {terminated}")
        
        if terminated or truncated:
            print(f"Episode Ended at step {i+1}. Final Reward: {reward:.4f}")
            obs, info = env.reset() 
            break

env.close()
```
