# 3-D Traffic Avoidance Demo

Two-level environment: traffic lanes on the **ground floor** and an
**elevated overpass** on floor 1 that the agent can use to cross safely.

```python
import time
import gymnasium as gym
from gymnasium.envs.registration import register
from environments import TrafficSim3DEnv
from globals import MAX_STEPS, SPEED

register(
    id="TrafficSim3D-StaticDisappear-v0",
    entry_point="environments:TrafficSim3DEnv",
    max_episode_steps=100,
    kwargs={
        "width": 16,
        "static_disappearing_mode": True,
        "disappear_prob": 0.15,
        "target_concentration_x": 10,
        "concentration_span": 3,
    },
)
```

```python
env = gym.make("TrafficSim3D-StaticDisappear-v0", render_mode="rgb_array")
obs, info = env.reset()
print("Mission:", env.unwrapped.mission)
print(f"Agent at floor {env.unwrapped.agent_pos[2]}, goal at floor {env.unwrapped.goal_pos[2]}")

total_reward = 0

for step in range(200):
    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(action)
    total_reward += reward

    pos = env.unwrapped.agent_pos
    print(
        f"Step {step:03d}  action={action}  "
        f"pos=({pos[0]},{pos[1]},F{pos[2]})  "
        f"reward={reward:.2f}  done={terminated}"
    )

    if terminated or truncated:
        break

env.close()
print(f"\nTotal reward: {total_reward:.2f}")
```

## Render a Snapshot

```python
import matplotlib.pyplot as plt

snapshot_env = TrafficSim3DEnv(
    render_mode="rgb_array",
    width=16,
    static_disappearing_mode=True,
    disappear_prob=0.15,
    target_concentration_x=10,
    concentration_span=3,
)
snapshot_env.reset()

img = snapshot_env.render()
plt.figure(figsize=(12, 5))
plt.imshow(img)
plt.axis("off")
plt.title("Traffic 3-D – Floor Views")
plt.show()
```
